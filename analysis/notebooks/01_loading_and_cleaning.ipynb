# 01 — Loading & Cleaning

Provenance + schema check for the requested run.

```python
import sys
from pathlib import Path as _P
_root = _P.cwd()
while not (_root / 'pyproject.toml').exists() and _root != _root.parent:
    _root = _root.parent
if str(_root) not in sys.path:
    sys.path.insert(0, str(_root))
import pandas as pd
run_id = 'exp-rq1-baseline'
from analysis.scripts.load_runs import load_results
df, meta = load_results(run_id)
display(df.head())
print('rows:', len(df))
expected = {'run_id','iteration','server_variant','defense_level','attack_id','boundary','event_type','success','latency_ms','tenant_pair','payload_sha256','timestamp'}
missing = expected - set(df.columns)
extra   = set(df.columns) - expected
assert not missing, f'missing columns: {missing}'
print('schema OK; extra columns (ignored):', sorted(extra))
```
